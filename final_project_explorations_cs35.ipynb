```python
import pandas as pd
import numpy as np
from sklearn.neural_network import MLPRegressor

# Preliminary Exploration: Converting a Melody into "Feature Windows"
# Example sequence: C -> D -> E -> F -> G -> A -> B -> C
melody = [60, 62, 64, 65, 67, 69, 71, 72] 

def prepare_melody_data(sequence, window_size=3):
    X, y = [], []
    for i in range(len(sequence) - window_size):
        X.append(sequence[i:i + window_size]) # The "context" notes
        y.append(sequence[i + window_size])    # The "target" next note
    return np.array(X), np.array(y)

# Creating the training set (e.g., use 3 notes to predict the 4th)
X_train, y_train = prepare_melody_data(melody)

print("Musical Feature Matrix (X):\n", X_train)
print("\nMusical Target Note (y):\n", y_train)

# Proof of Concept: Can the NNet 'hear' the pattern?
nn_composer = MLPRegressor(hidden_layer_sizes=(15, 15), max_iter=1000, random_state=42)
nn_composer.fit(X_train, y_train)
print("\nStatus: Preliminary NNet training on melodic data successful.")
```

On your MLPRegressor vs MLPClassifier question: I'd lean toward Classifier. Notes are discrete categories (MIDI pitch 60 ≠ "halfway between 59 and 61" in any musical sense — they're different notes). A regressor will happily predict pitch 63.7, which isn't a note. Classifier also gives you probability distributions you can sample from, which produces more interesting/varied output than always picking the argmax. Sampling from the top-k predictions is how you get "creativity" without a big model.

"Pi music" can map the digits of pi (or any number sequence) to musical notes. This would be a good baseline because it is:
- deterministic but sounds pseudo-random
- trivial to implement (1 function)

Baseline 1: Uniformly random notes (sounds like noise)
Baseline 2: π-digit or frequency-weighted random (slightly better — uses pitch distribution from training data)
Your model: MLP with rolling-window features (should capture local melodic patterns)
Evaluation: Accuracy on next-note prediction + qualitative listening
